**AI RESPONSE** (paste it here, or summarize the key points): In scikit-learn:

**`fit()` = learn from the data**
It calculates and stores information from the dataset.

Example: `StandardScaler().fit(X_train)` learns the mean and standard deviation of each column.

**`transform()` = apply what was learned**
It uses the information learned during `fit()` to change the data.

Example: after fitting a scaler on `X_train`, you use:

```python
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

This applies the same scaling rules to both datasets.

**`fit_transform()` = learn and apply in one step**
It does `fit()` and `transform()` together. Scikit-learn says this is often more convenient and efficient for transforming training data. ([Scikit-learn][1])

```python
X_train_scaled = scaler.fit_transform(X_train)
```

This is basically like:

```python
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
```

Scikit-learn’s `TransformerMixin` defines `fit_transform` as a method that delegates to `fit` and `transform`. ([Scikit-learn][2])

**When to use each:**

Use **`fit()`** when you only want the object to learn patterns/statistics from the data.

Use **`transform()`** when the object has already been fitted, and you want to apply the same rule to training, validation, or test data.

Use **`fit_transform()`** on the **training data only** when you want to learn and transform at the same time.

Most important rule:

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Do **not** use `fit_transform()` on the test data, because that would let the test data influence the preprocessing and cause data leakage.

[2]: https://scikit-learn.org/stable/modules/generated/sklearn.base.TransformerMixin.html?utm_source=chatgpt.com "TransformerMixin"
